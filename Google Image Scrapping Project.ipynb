```python
Soup = BeautifulSoup(response.content,'html.parser')
```

```python
# Scrapping the Image Tags(Extract the image Links)
images_tags = Soup.find_all("img")
```

```python
# Now we Remove the Top short image which shows us which types of images we need which is in the short image type so we Remove this.
del images_tags[0]
```

```python
"""Now we Source code So we Extrace the src code means 
Source Code by using the For Loop"""
img_data_mongo = []
for i in images_tags:
    image_url = i['src']
    image_data = requests.get(image_url).content
    my_dict = {'index':image_url,'image':image_data}
    img_data_mongo.append(my_dict)

    with open(os.path.join(save_dir,f"{query}_{images_tags.index(i)}.jpg"),"wb") as f:
        f.write(image_data)
```
